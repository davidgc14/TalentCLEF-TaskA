# Graficas de evaluacion final

```python
import pandas as pd

df = pd.read_csv('models/results/evaluation_results.csv')
df
```

```python
# hacer grafica de puntos donde el eje x sea num_params y el eje y sea map_es_es. Diferencia por colores la fuente (source)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(df['num_params'], df['map_es_es'], c=df['source'].map({'hugging_face': 'blue', 'spanish': 'green', 'mteb': 'red'}), alpha=0.7)
plt.xlabel('Número de parámetros')
plt.ylabel('MAP (es-es)')
plt.xscale('log')  # Escala logarítmica para el eje x
plt.grid(True)
# mostrar Leyenda 
handles = [plt.Line2D([0], [0], marker='o', color='w', label='Hugging Face', markerfacecolor='blue', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='Spanish', markerfacecolor='green', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='MTEB', markerfacecolor='red', markersize=10)]
plt.legend(handles=handles)
plt.show()
```

```python
# ahora que se pinten en distintos colores los monolinuges y los multilingues
plt.figure(figsize=(10, 6))
plt.scatter(df['num_params'], df['map_es_es'], c=df['languages'].map({'monolingual': 'orange', 'multilingual': 'purple'}), alpha=0.7)
plt.xlabel('Número de parámetros')
plt.ylabel('MAP (es-es)')
plt.xscale('log')  # Escala logarítmica para el eje x
plt.grid(True)
# mostrar Leyenda
handles = [plt.Line2D([0], [0], marker='o', color='w', label='Monolingual', markerfacecolor='orange', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='Multilingual', markerfacecolor='purple', markersize=10)]
plt.legend(handles=handles)
plt.show()
```
