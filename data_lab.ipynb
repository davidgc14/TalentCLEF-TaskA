# Spanish models

```python
from huggingface_hub import list_models

# Lista de autores
autores = ["PlanTL-GOB-ES", "dccuchile"]
librerias = ["transformers", "sentence-transformers"]

# Recopilar modelos de todos los autores
todos_los_modelos = []

for libreria in librerias:
    for autor in autores:
        modelos = list(list_models(
            author=autor,
            library=libreria,
        ))
        todos_los_modelos.extend(modelos)

# Ordenar por ID
todos_los_modelos = sorted(todos_los_modelos, key=lambda x: x.id)

# Mostrar los modelos
print(f"Total de modelos encontrados: {len(todos_los_modelos)}\n")
for model in todos_los_modelos:
    print(model.id)
```

```python

model_list = [model.id for model in todos_los_modelos if 'finetuned' not in model.id]
model_list
```

```python
model_list.extend([
    "hiiamsid/sentence_similarity_spanish_es",
    "hackathon-pln-es/paraphrase-spanish-distilroberta",
    "jinaai/jina-embeddings-v2-base-es",
    "jaimevera1107/all-MiniLM-L6-v2-similarity-es",
    "jaimevera1107/roberta-similarity-es",
    "bertin-project/bertin-roberta-base-spanish",
    "jfarray/Model_dccuchile_bert-base-spanish-wwm-uncased_100_Epochs"
])
```

```python
# exportar lista a txt
with open('es_models.txt', 'w') as f:
    for item in model_list:
        f.write("%s\n" % item)
```

# Unificar Ranking

```python
import pandas as pd

df_hf = pd.read_csv('models/results/hugging_face/hf_evaluation_results.csv')
df_es = pd.read_csv('models/results/spanish/es_evaluation_results.csv')
df_mteb = pd.read_csv('models/results/mteb/mteb_evaluation_results.csv')

df_mteb
```

```python
df_hf['source'] = 'hugging_face'
df_es['source'] = 'spanish'
df_mteb['source'] = 'mteb'
df_combined = pd.concat([df_hf, df_es, df_mteb], ignore_index=True)
# ordenar por 'MAP'
df_combined = df_combined.sort_values(by='map_es_es', ascending=False)
df_combined.reset_index(drop=True, inplace=True)
df_combined
```

```python
df_combined[['model_name', 'num_params', 'languages', 'map_es_es', 'source']].head(20)
```

```python
# hacer grafica de puntos donde el eje x sea num_params y el eje y sea map_es_es. Diferencia por colores la fuente (source)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(df_combined['num_params'], df_combined['map_es_es'], c=df_combined['source'].map({'hugging_face': 'blue', 'spanish': 'green', 'mteb': 'red'}), alpha=0.7)
plt.xlabel('Número de parámetros')
plt.ylabel('MAP (es-es)')
plt.xscale('log')  # Escala logarítmica para el eje x
plt.grid(True)
# mostrar Leyenda 
handles = [plt.Line2D([0], [0], marker='o', color='w', label='Hugging Face', markerfacecolor='blue', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='Spanish', markerfacecolor='green', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='MTEB', markerfacecolor='red', markersize=10)]
plt.legend(handles=handles)
plt.show()
```

```python
# ahora que se pinten en distintos colores los monolinuges y los multilingues
plt.figure(figsize=(10, 6))
plt.scatter(df_combined['num_params'], df_combined['map_es_es'], c=df_combined['languages'].map({'monolingual': 'orange', 'multilingual': 'purple'}), alpha=0.7)
plt.xlabel('Número de parámetros')
plt.ylabel('MAP (es-es)')
plt.xscale('log')  # Escala logarítmica para el eje x
plt.grid(True)
# mostrar Leyenda
handles = [plt.Line2D([0], [0], marker='o', color='w', label='Monolingual', markerfacecolor='orange', markersize=10),
           plt.Line2D([0], [0], marker='o', color='w', label='Multilingual', markerfacecolor='purple', markersize=10)]
plt.legend(handles=handles)
plt.show()
```
